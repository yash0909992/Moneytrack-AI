```python
# ITEMS CATEGORIZATION  
def categorize_expense(description):

    description = str(description).lower()

    if any(word in description for word in [
        "swiggy", "zomato", "restaurant", "food"
    ]):
        return "Food"

    elif any(word in description for word in [
        "uber", "ola", "rapido", "metro", "bus",
        "train", "petrol", "fuel"
    ]):
        return "Transport"

    elif any(word in description for word in [
        "amazon", "flipkart", "myntra", "shopping"
    ]):
        return "Shopping"

    elif any(word in description for word in [
        "netflix", "spotify", "movie", "prime"
    ]):
        return "Entertainment"

    elif any(word in description for word in [
        "rent", "electricity", "internet", "water"
    ]):
        return "Bills & Housing"

    elif any(word in description for word in [
        "pharmacy", "hospital", "medical", "medicine"
    ]):
        return "Healthcare"

    elif any(word in description for word in [
        "salary", "income", "bonus"
    ]):
        return "Income"

    else:
        return "Other"
```

```python
import numpy as np
import pandas as pd

np.random.seed(42)

months = pd.date_range(
    start="2025-01-01",
    periods=12,
    freq="MS"
)

base_expense = 20000

expenses = []

for i in range(12):
    trend = i * 800
    variation = np.random.randint(-1500, 1500)

    expense = base_expense + trend + variation

    expenses.append(expense)

monthly_ml = pd.DataFrame({
    "Month": months,
    "Expense": expenses
})

monthly_ml
```

```python
monthly_ml["Month_Number"] = range(
    1,
    len(monthly_ml) + 1
)

monthly_ml
```

```python
from sklearn.linear_model import LinearRegression

X = monthly_ml[["Month_Number"]] 
y = monthly_ml["Expense"]

model = LinearRegression()

model.fit(X, y)
```

```python
from sklearn.metrics import mean_absolute_error, r2_score

y_pred = model.predict(X)

mae = mean_absolute_error(y, y_pred) 
r2 = r2_score(y, y_pred) 

print(f"MAE: ₹{mae:,.2f}") 
print(f"R² Score: {r2:.2f}")  
```

```python
next_month = pd.DataFrame({
    "Month_Number": [13]
})

predicted_expense = model.predict(next_month)[0] 

print(
    f"Predicted Next Month Expense: ₹{predicted_expense:,.2f}"
)
```

```python
monthly_ml["Predicted_Expense"] = model.predict(X)

fig = px.line(
    monthly_ml,
    x="Month",
    y=["Expense", "Predicted_Expense"],
    markers=True,
    title="Actual vs Predicted Monthly Expenses"
)

fig
```

```python
import numpy as np
import pandas as pd

np.random.seed(42)

n = 200

income = np.random.randint(20000, 100001, n)
expense = np.random.randint(10000, 90001, n)

expense = np.minimum(expense, income - 1000)

savings = income - expense

savings_rate = (savings / income) * 100

savings_health_df = pd.DataFrame({
    "Income": income,
    "Expense": expense,
    "Savings": savings,
    "Savings_Rate": savings_rate
})

savings_health_df.head()
```

```python
def savings_health(rate):

    if rate < 10:
        return "Poor"

    elif rate < 20:
        return "Average"

    elif rate < 30:
        return "Good"

    else:
        return "Excellent"
```

```python
savings_health_df["Health"] = (
    savings_health_df["Savings_Rate"]
    .apply(savings_health)
)

savings_health_df.head()
```

```python
savings_health_df["Health"].value_counts()
```

```python
X = savings_health_df[
    ["Income", "Expense", "Savings", "Savings_Rate"]
]

y = savings_health_df["Health"]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)
```

```python
from sklearn.tree import DecisionTreeClassifier

model2 = DecisionTreeClassifier(
    max_depth=3,
    random_state=42
)
```

```python
model2.fit(
    X_train,
    y_train
)
```

```python
y_pred2 = model2.predict(X_test)

y_pred2[:10]
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(
    y_test,
    y_pred2 
)

print(f"Model 2 Accuracy: {accuracy:.2%}")
```

```python
from sklearn.metrics import classification_report

print(
    classification_report(
        y_test,
        y_pred2,
        zero_division=0
    )
)
```

```python
def financial_advisor(
    income,
    expenses,
    savings,
    savings_rate,
    highest_category
):

    advice = []

    # Check savings
    if savings_rate < 10:
        advice.append(
            "Your savings rate is low. "
            "Focus on reducing unnecessary expenses."
        )

    elif savings_rate < 20:
        advice.append(
            "Your savings rate is moderate. "
            "Try to increase your monthly savings gradually."
        )

    elif savings_rate < 30:
        advice.append(
            "Your savings rate is healthy. "
            "Continue maintaining your saving habit."
        )

    else:
        advice.append(
            "Your savings rate is strong. "
            "Continue maintaining disciplined spending."
        )

    # Check highest spending category
    if highest_category:
        advice.append(
            f"Your highest spending category is "
            f"{highest_category}. Review this category "
            f"for unnecessary expenses."
        )

    # Check expense ratio
    expense_ratio = (expenses / income) * 100

    if expense_ratio > 80:
        advice.append(
            "Your expenses are using a large portion "
            "of your income. Consider setting a spending limit."
        )

    elif expense_ratio < 50:
        advice.append(
            "Your expenses are relatively controlled "
            "compared with your income."
        )

    return advice
```

```python
def goal_planner(
    goal_amount,
    current_savings,
    monthly_savings,
    months
):
    
    remaining_amount = goal_amount - current_savings
    
    if remaining_amount <= 0:
        return {
            "status": "Goal Already Achieved",
            "message": "You have already reached your savings goal."
        }
    
    required_monthly_saving = (
        remaining_amount / months
    )
    
    if monthly_savings >= required_monthly_saving:
        status = "Goal Achievable"
        
        message = (
            f"You need to save ₹{required_monthly_saving:,.2f} "
            f"per month. Your current monthly saving of "
            f"₹{monthly_savings:,.2f} is sufficient."
        )
        
    else:
        status = "Needs Improvement"
        
        message = (
            f"You need to save ₹{required_monthly_saving:,.2f} "
            f"per month. Your current monthly saving of "
            f"₹{monthly_savings:,.2f} is not sufficient."
        )
    
    return {
        "status": status,
        "required_monthly_saving": required_monthly_saving,
        "message": message
    }
```

```python
goal_result = goal_planner(
    goal_amount=100000,
    current_savings=20000,
    monthly_savings=25000,
    months=4
)

goal_result
```

```python
print("🎯 Goal Status:", goal_result["status"])

if "required_monthly_saving" in goal_result:
    print(
        f"💰 Required Monthly Saving: "
        f"₹{goal_result['required_monthly_saving']:,.2f}"
    )

print("💡", goal_result["message"])
```

```python
def what_if_simulator(
    income,
    expenses,
    savings,
    change_type,
    change_percent
):

    if change_type == "expense_reduction":

        reduction = expenses * (change_percent / 100)

        new_expenses = expenses - reduction

        new_savings = income - new_expenses

        savings_change = new_savings - savings

        return {
            "Current Expenses": expenses,
            "New Expenses": new_expenses,
            "Extra Savings": savings_change
        }

    elif change_type == "income_increase":

        increase = income * (change_percent / 100)

        new_income = income + increase

        new_savings = new_income - expenses

        savings_change = new_savings - savings

        return {
            "Current Income": income,
            "New Income": new_income,
            "Extra Savings": savings_change
        }
```

```python
result = what_if_simulator(
    income=50000,
    expenses=40000,
    savings=10000,
    change_type="expense_reduction",
    change_percent=20
)

result
```

```python
print("🔮 WHAT-IF SIMULATION")
print("----------------------")

for key, value in result.items():
    print(f"{key}: ₹{value:,.2f}")
```

```python
def category_what_if(
    expense_by_category,
    total_expenses,
    category,
    reduction_percent
):

    current_category_expense = expense_by_category[category]

    reduction = (
        current_category_expense *
        reduction_percent / 100
    )

    new_category_expense = (
        current_category_expense - reduction
    )

    new_total_expenses = (
        total_expenses - reduction
    )

    extra_savings = reduction

    return {
        "Category": category,
        "Current Category Expense": current_category_expense,
        "Reduction": reduction,
        "New Category Expense": new_category_expense,
        "New Total Expenses": new_total_expenses,
        "Extra Savings": extra_savings
    }
```

```python
result = category_what_if(
    expense_by_category=expense_by_category,
    total_expenses=total_expenses,
    category="Shopping",
    reduction_percent=20
)

result
```
